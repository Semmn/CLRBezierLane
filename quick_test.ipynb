```python
import torch
sd = torch.load("work_dirs/clrbezier/culane/clrbezier_anchored_o2m_r34/run1/epoch_36.pth", map_location="cpu")["state_dict"]
lk = next(k for k in sd if k.endswith("prior_logits"))
cp = torch.sigmoid(sd[lk] + 0.1 * sd[lk.replace("prior_logits", "prior_delta")])
x  = cp.view(-1, 4, 2)[..., 0]                      # control-point x, [K, 4]
a, b = x[:,0] - 2*x[:,1] + x[:,2], x[:,1] - 2*x[:,2] + x[:,3]
bend = (12 * (a**2 + a*b + b**2)).sqrt()            # curvature norm, x''(t) = 6[(1-t)a + t b]
print(f"{x.shape[0]} anchors | |a| {a.abs().mean():.5f}  |b| {b.abs().mean():.5f}")
print(f"bend: mean {bend.mean()*800:.2f} px   max {bend.max()*800:.2f} px")
print(f"delta moved {(0.1*sd[lk.replace('prior_logits','prior_delta')]).abs().mean():.5f} logits")
```
